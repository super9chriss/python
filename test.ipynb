```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
# 1. Generate numerical data with NumPy
x = np.linspace(0, 10, 100)
y = np.sin(x)
# 2. Render plot with Matplotlib
plt.figure(figsize=(5, 2.2))
plt.plot(x, y, label='Sine Wave', color='#0284c7')
plt.title("VS Code + Jupyter Setup Complete!")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.legend()
plt.show()
# 3. Print library versions
print(f"NumPy Version: {np.__version__}")
print(f"SciPy Version: {sp.__version__}")
```
