# 1st Assignment — Sensor Data Analysis Tool

## Objective

Develop a Python program that simulates sensor readings, analyzes the generated measurements, and visualizes the results.

You will combine the Python fundamentals from Class 1, modules and file handling from Class 2, and NumPy and Matplotlib from Class 3.

**Important:** Use functions and modules only. Do not use classes or object-oriented programming yet.

## 1. User input

When the program starts, ask the user to enter:

- Sensor type (for example, temperature or light).
- Number of readings to generate.
- Minimum sensor value.
- Maximum sensor value.
- Alert threshold.

## 2. Generate sensor readings

Use Python's `random` module to generate the requested number of simulated sensor readings.

Requirements:

- Store the readings in a list or NumPy array.
- Use the minimum and maximum values entered by the user.
- Use a function to generate the readings.
- Make the results reproducible with `random.seed()`.

## 3. Analyze the data with NumPy

Calculate and display:

- Number of readings.
- Mean.
- Minimum.
- Maximum.
- Standard deviation.
- Number of readings above the alert threshold.

Identify and display the readings that exceed the threshold.

## 4. Visualize the data with Matplotlib

Create two graphs:

1. **Line plot:** Show sensor readings against sample number. Add a horizontal line indicating the alert threshold.
2. **Histogram:** Show the distribution of the readings.

Both graphs should include appropriate titles, axis labels, and a grid. Add a legend where appropriate.

## 5. Save the data

Save the generated readings to a CSV file.

The CSV file should contain at least:
- Sample number.
- Sensor reading.

Use Python's `csv` module.

## 6. Choice

Allow the user to choose between generating new sensor readings and loading readings from an existing CSV file.

If the user loads existing data, analyze and plot that data without generating new readings.

## 7. Program structure

Write the program in one main Python file named `sensor_monitor.py`.

Your program must use functions such as:

- `get_user_input()`
- `generate_readings()`
- `calculate_statistics()`
- `find_outliers()`
- `save_readings()`
- `plot_readings()`
- `plot_histogram()`
- `main()`

You may use different function names, provided that the program is clearly structured.

Use the following entry point:

```python
if __name__ == "__main__":
    main()
```

## 8. Error handling

Your program must handle invalid user input:

- Text entered where a number is expected.
- A zero or negative number of readings.
- A minimum value that is greater than or equal to the maximum.
- A problem saving the CSV file.

Display helpful error messages instead of allowing the program to fail unexpectedly.

## 8. GitHub submission

Submit the project through GitHub.

Your repository should contain:

Your `requirements.txt` should indicate the programm dependencies.

Make meaningful Git commits during development and push your completed project to GitHub, not just a single last minute upload..
